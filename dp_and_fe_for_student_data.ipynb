# หาค่าเฉลี่ยของอายุและนำค่าเฉลี่ยที่ได้ไปใส่แทนที่อายุที่เป็นค่าว่าง(NaN)

# แทนที่ค่าว่าง(NaN)ในคอลัมน์คณะด้วยฐานนิยมของคอลัมน์คณะ(เจอคณะใดคณะหนึ่งเยอะที่สุดในคอลัมน์)ซึ่งฐานนิยมคือ Science ก็จะนำ Science ไปแทนที่ค่าว่างในคอลัมน์คณะ

```python
df['คณะ'].fillna(df['คณะ'].mode()[0], inplace=True)
```

```python
df
```

# แสดงคอลัมน์คณะโดยที่ทำให้คณะเดียวกันที่เขียนไม่เหมือนกันให้เขียนเหมือนกัน

```python
def standardize_faculty(name):
    if name in ['วิทยาศาสตร์', 'วิทย์', 'Science', 'คณะวิทยาศาสตร์','Sci']:
        return 'วิทยาศาสตร์'
    elif name in ['ศิลปศาสตร์', 'ศิลป์', 'Arts', 'Liberal Arts']:
        return 'ศิลปศาสตร์'
    elif name in ['วิศวะ','วิศวกรรม','Engineering','วิศวฯ','วิศวกรรมศาสตร์']:
        return 'วิศวกรรมศาสตร์'
    else:
        return name
```

```python
df['คณะ_มาตรฐาน'] = df['คณะ'].apply(standardize_faculty)
```

```python
df['คณะ_มาตรฐาน']
```

# แสดงข้อมูลที่อยู่ในควอไทล์แรกกับควอไทล์ที่สาม

```python
Q1 = df['อายุ'].quantile(0.25)
Q3 = df['อายุ'].quantile(0.75)
```

```python
Q1, Q3
```

# แสดงข้อมูลที่เป็นส่วนต่างระหว่างควอไทล์แรกกับควอไทล์ที่สาม

```python
IQR = Q3 - Q1
```

```python
IQR
```

# แสดงขอบเขตล่างกับขอบเขตบน สำหรับวิเคราะห์หาค่าผิดปกติ

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
lower_bound, upper_bound
```

# แสดงข้อมูลอายุที่ไม่ต่ำกว่าขอบเขตล่างและไม่สูงกว่าขอบเขตบน


```python
df_no_outliers = df[(df['อายุ'] >= lower_bound) & (df['อายุ'] <= upper_bound)]
```

```python
df_no_outliers
```

# แสดงข้อมูลอายุโดยถ้าเป็นTrue คือ ค่าอายุมากกว่าหรือเท่ากับขอบเขตล่าง แต่ถ้าเป็นFalse คือ ค่าอายุน้อยกว่าขอบเขตล่าง

```python
df['อายุ'] >= lower_bound
```

# แสดงข้อมูลอายุโดยถ้าเป็นTrue คือ ค่าอายุน้อยกว่าหรือเท่ากับขอบเขตบน แต่ถ้าเป็นFalse คือ ค่าอายุมากกว่ากว่าขอบเขตบน

```python
df['อายุ'] <= upper_bound
```

# แสดงข้อมูลอายุโดยที่ถ้าเป็น True คือ ค่าอายุมากกว่าหรือเท่ากับขอบเขตล่างและน้อยกว่าหรือเท่ากับขอบเขตบน แต่ถ้าเป็น False คือ ค่าอายุจะน้อยกว่าขอบเขตล่างและมากกว่าขอบเขตบน

```python
(df['อายุ'] >= lower_bound) & (df['อายุ'] <= upper_bound)
```

#ทำการแปลงข้อมูลในคอลัมน์'วันที่ลงทะเบียน'ให้กลายเป็นชนิดข้อมูลdatetimeของpandasก่อน เพื่อที่จะสามารถใช้functionวัน-เวลาของpandasได้

# สร้างคอลัมน์ใหม่ชื่อว่า'วันในสัปดาห์'และใส่ข้อมูลลงในคอลัมน์โดยข้อมูลเอามาจากคอลัมน์'วันที่ลงทะเบียน'ดูจากวันที่และนำมาแปลงเป็นวันเป็นภาษาอังกฤษ

#สร้างคอลัมน์ใหม่ชื่อ'จำนวนคาบ'ข้อมูลที่นำมาใส่ในคอลัมน์มาจากการหาชั่วโมงเรียนต่อสัปดาห์ โดยจะทำข้อมูลให้เป็นช่วงๆ เช่น ชั่วโมงเรียนต่อสัปดาห์ คือ 15 อยู่ในช่วงที่กำหนดไว้คือ 8-16 ดังนั้นก็จะได้จำนวนคาบคือ 2 วัน ก็จะนำ 2 วัน ได้ไปใส่ลงในคอลัมน์'จำนวนคาบ'

```python
df['จำนวนคาบ'] = pd.cut(df['ชั่วโมงเรียน_ต่อสัปดาห์'],
                                  bins=[0, 8, 16, 24, 32, 40],
                                  labels=['1 วัน', '2 วัน', '3 วัน', '4 วัน', '5 วัน'])
```

```python
df['จำนวนคาบ']
```

```python
df
```
